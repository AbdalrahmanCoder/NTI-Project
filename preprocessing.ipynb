```python
import pandas as pd

import numpy as np

from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv("Loan Approval Prediction.csv")

df.head()
```

```python
print(df.shape)

df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

df.shape
```

```python
df.describe()
```

```python
X = df.drop("loan_status", axis=1)
y = df["loan_status"]
```

```python
#puts numerical features on a similar scale

numerical_columns = X.select_dtypes(include=["int64", "float64"]).columns
scaler = StandardScaler()
X[numerical_columns] = scaler.fit_transform(X[numerical_columns])
X.head()
```

```python
# Combine into one dataset
processed_df = pd.concat([X, y], axis=1)
processed_df.head()
```

```python
print(processed_df.shape)
print(processed_df.columns.tolist())
```

```python
# Save the processed dataset
processed_df.to_csv("processed_loan_approval.csv", index=False)

# Save the scaler
import pickle

with open("loan_approval_scaler.pkl", "wb") as f:
    pickle.dump(scaler, f)

print("Dataset 1 preprocessing saved!")
```

## Dataset 2

```python
df2=pd.read_csv("loan_data.csv")
```

```python
df2.head()
```

```python
print(df2.shape)

df2.info()
```

```python
df2.isnull().sum()
```

```python
df2.duplicated().sum()
```

```python
# Find columns that contain text/categorical values
categorical_columns = df2.select_dtypes(include=["object"]).columns

print("Categorical columns:")
print(categorical_columns)
```

```python
label_encoder = LabelEncoder()

for column in categorical_columns:
    df2[column] = label_encoder.fit_transform(df2[column])

df2.head()
```

```python
# Check the data types after encoding
df2.info()
```

```python
df2.describe()
```

```python
X = df2.drop("loan_status", axis=1)
y = df2["loan_status"]
```

```python
numerical_columns = X.columns

scaler = StandardScaler()

X[numerical_columns] = scaler.fit_transform(X[numerical_columns])

X.head()
```

```python
# Add the target column back to the processed features
processed_df2 = pd.concat([X, y], axis=1)

# Display the final preprocessed dataset
processed_df2.head()
```

```python
print(processed_df2.shape)
processed_df2.info()

processed_df2.to_csv("processed_loan_data.csv", index=False)

# Save the scaler and encoders
assets = {
    "scaler": scaler,
    "encoders": label_encoder,
    "feature_names": list(X.columns)
}

with open("loan_data_preprocessing.pkl", "wb") as f:
    pickle.dump(assets, f)

print("Dataset 2 preprocessing saved!")
```
